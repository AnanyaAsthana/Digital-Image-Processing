```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
```

```python
# Create a synthetic grayscale image (low contrast)
synthetic = np.tile(np.linspace(50, 150, 256), (256, 1)).astype(np.uint8)

# Histogram Equalization
equalized = cv2.equalizeHist(synthetic)

# Display
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.title("Original Synthetic Image")
plt.imshow(synthetic, cmap='gray')
plt.axis('off')

plt.subplot(1,2,2)
plt.title("Histogram Equalized Image")
plt.imshow(equalized, cmap='gray')
plt.axis('off')
plt.show()
```

```python
A = np.zeros((1024,1024), dtype=np.uint8)

for i in range(0, 1024, 256):
    A[i:i+128, :] = 255
```

```python
B = np.zeros((1024,1024), dtype=np.uint8)

for j in range(0, 1024, 256):
    B[:, j:j+128] = 255
```

```python
# Convert to float for safe arithmetic
A_f = A.astype(np.float32)
B_f = B.astype(np.float32)

# a) Addition
add_img = cv2.add(A, B)

# b) Subtraction
sub_img = cv2.subtract(A, B)

# c) Multiplication (normalize)
mul_img = cv2.multiply(A_f/255, B_f/255)

# d) Averaging
avg_img = cv2.addWeighted(A, 0.5, B, 0.5, 0)

# Display results
titles = ["A", "B", "A + B", "A - B", "A * B", "Average"]
images = [A, B, add_img, sub_img, mul_img, avg_img]

plt.figure(figsize=(12,8))
for i in range(6):
    plt.subplot(2,3,i+1)
    plt.title(titles[i])
    plt.imshow(images[i], cmap='gray')
    plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
img = np.ones((256,256), dtype=np.uint8) * 255

# Black box of size 58x58
start = (256 - 58)//2
end = start + 58

img[start:end, start:end] = 0

plt.figure(figsize=(4,4))
plt.title("White Image with Black Box")
plt.imshow(img, cmap='gray')
plt.axis('off')
plt.show()
```
