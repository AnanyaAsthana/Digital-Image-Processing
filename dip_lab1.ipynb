```python
# Split image into Blue, Green, Red channels
B, G, R = cv2.split(img)
```

```python
print("Blue Plane")
cv2_imshow(B)

print("Green Plane")
cv2_imshow(G)

print("Red Plane")
cv2_imshow(R)
```

```python
cv2.imwrite("blue_plane.png", B)
cv2.imwrite("green_plane.png", G)
cv2.imwrite("red_plane.png", R)
```

```python
import cv2
from google.colab.patches import cv2_imshow

# Read grayscale channel images
R = cv2.imread("red_plane.png", cv2.IMREAD_GRAYSCALE)
G = cv2.imread("green_plane.png", cv2.IMREAD_GRAYSCALE)
B = cv2.imread("blue_plane.png", cv2.IMREAD_GRAYSCALE)
```

```python
# Merge channels (B, G, R order)
color_img = cv2.merge((B, G, R))
```

```python
print("Reconstructed Color Image")
cv2_imshow(color_img)
```

```python
cv2.imwrite("reconstructed_color.png", color_img)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
```

```python
print("Grayscale Image")
cv2_imshow(gray)
```

```python
cv2.imwrite("grayscale.png", gray)
```

```python
# Fixed threshold
_, binary = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)
```

```python
print("Binary Image")
cv2_imshow(binary)
```

```python
cv2.imwrite("binary.png", binary)
```

```python
half = cv2.resize(img, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)
```

```python
quarter = cv2.resize(img, None, fx=0.25, fy=0.25, interpolation=cv2.INTER_LINEAR)
```

```python
print("Half Size Image")
cv2_imshow(half)

print("Quarter Size Image")
cv2_imshow(quarter)
```

```python
cv2.imwrite("half.png", half)
cv2.imwrite("quarter.png", quarter)
```

```python
(h, w) = img.shape[:2]
center = (w // 2, h // 2)
```

```python
M45 = cv2.getRotationMatrix2D(center, 45, 1.0)
rot_45 = cv2.warpAffine(img, M45, (w, h))
```

```python
M90 = cv2.getRotationMatrix2D(center, 90, 1.0)
rot_90 = cv2.warpAffine(img, M90, (w, h))
```

```python
M10 = cv2.getRotationMatrix2D(center, 10, 1.0)
rot_10 = cv2.warpAffine(img, M10, (w, h))
```

```python
print("45 Degree Rotation")
cv2_imshow(rot_45)

print("90 Degree Rotation")
cv2_imshow(rot_90)

print("10 Degree Rotation")
cv2_imshow(rot_10)
```
